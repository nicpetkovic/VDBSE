```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Choose the power profile to analyze

```python
file_name = "01_powerprofile_10degrees_0capacityreduction.csv"
```

```python
file_path = f"../../data/power_profile_tests/{file_name}"
df = pd.read_csv(file_path)
```

```python
df.describe()
```

```python
df.head()
```

```python
time = np.arange(0,len(df['SoC']),1)
SoC = np.array(df['SoC'])
Vocv = np.array(df['Vocv'])
```

```python
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

axs[0, 0].plot(time, SoC, label="SoC", color="orange")
axs[0, 0].set_xlabel("Time")
axs[0, 0].set_ylabel("State of Charge (SoC)")
axs[0, 0].set_title("SoC vs Time")
axs[0, 0].legend()
axs[0, 0].grid()

axs[0, 1].plot(time, Vocv, label="Vocv", color="blue")
axs[0, 1].set_xlabel("Time")
axs[0, 1].set_ylabel("Voltage openc circuit (Vocv)")
axs[0, 1].set_title("Vocv vs Time")
axs[0, 1].legend()
axs[0, 1].grid()

axs[1, 0].plot(time, df["I"], color="red")
axs[1, 0].set_title("Current vs Time")
axs[1, 0].set_xlabel("Time")
axs[1, 0].set_ylabel("Current (A)")
axs[1, 0].grid()

axs[1, 1].plot(time, df["V"], color="m")
axs[1, 1].set_title("Voltage")
axs[1, 1].set_xlabel("Time")
axs[1, 1].set_ylabel("Voltage (V)")
axs[1, 1].grid()

plt.tight_layout()
plt.show()
```
