```python
import scipy.io
import numpy as np
import pandas as pd
import os
```

```python
mat_file_path = "../data_matlab/power_profile_tests/01_powerprofile_10degrees_0capacityreduction.mat"
mat_contents = scipy.io.loadmat(mat_file_path)

mat_contents.keys()
```

```python
variables_info = {key: type(mat_contents[key]) for key in mat_contents if not key.startswith("__")}
variables_info
```

```python
variables_shapes = {key: mat_contents[key].shape for key in mat_contents if isinstance(mat_contents[key], np.ndarray)}
variables_shapes
```

```python
data = {
    "C": mat_contents["C"].flatten(),
    "I": mat_contents["I"].flatten(),
    "Rp": mat_contents["Rp"].flatten(),
    "Rs": mat_contents["Rs"].flatten(),
    "SoC": mat_contents["SoC"].flatten(),
    "V": mat_contents["V"].flatten(),
    "Vocv": mat_contents["Vocv"].flatten(),
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df.describe()
```

```python
folder_path = "../data/power_profile_tests"
file_name = "01_powerprofile_10degrees_0capacityreduction.csv"

os.makedirs(folder_path, exist_ok=True)

csv_file_path = os.path.join(folder_path, file_name)
df.to_csv(csv_file_path, index=False)

print(f"CSV file saved at: {csv_file_path}")
```
